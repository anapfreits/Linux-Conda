## Se localizando dentro do terminal

É muito importante que você tenha certeza de qual o seu local atual dentro do terminal para que as ações que voce pretende tomar dentro daquele ambiente sejam as mais corretas possivel. Saber de informações como: qual o nome seu usuário, nome da sua maquina e a sua localização atual é de extrema importância. Em seguida, voce verá como descobrir essas informações.

```bash
# Antes de tudo a funcao 'echo' é utilizada para imprimir informações que não serão interpretadas como códigos no terminal do usuário
# A função 'whoami' indica qual seu usuário atual
echo "Usuario atual:"
whoami

# A função 'hostname' indica qual a sua máquina
echo "Nome da maquina:"
hostname

# A função 'pwd' vai te informar qual o seu diretório atual e como chegar até ele
echo "Diretorio atual:"
pwd

# Indica o seu diretório pessoal
echo "Diretorio pessoal:"
echo "$HOME"
```

## Comando básicos do terminal linux

Você vai precisar desses comandos para se movimentar e desenvolver tarefas básicas dentro de um servidor. 
A partir da sua localização atual é importante que voce saiba para onde está indo e como voltar, por isso veja bem como listar os diretórios da sua maquina: 

```bash
# O comando 'ls' lista os diretórios a partir da pasta indicada 
echo "listar os diretorios"
ls

# Há algumas funções interessantes do ls como -al e -lh 
# O -al lista ate os arquivos ocultos do diretorio
ls -al 

# Já o -lh ele vai te mostrar o tamanho e as permissoes dos arquivos/diretorios 
ls -lh 

# Para se mover entre os diretorios basta usar 'cd' (change directory)
cd /caminho/do/diretorio

# Para criar um novo diretorio use 
mkdir nome_do_diretorio

# Para excluir um diretorio
rmdir nome_do_diretorio

# ou 
rm -r nome_do_diretorio
```

## Manipulação de arquivos

Após entender como acessar os diretorios da sua máquina, entenda como manipular os arquivos

```bash
# Crie um arquivo qualquer - esse arquivo pode ter qualquer extensão (.txt, .csv, .sh, etc)
nano arquivo.txt

# Com esse arquivo em mãos vc pode editá-lo usando o mesmo comando 'nano'
nano arquivo.txt

# Copia-lo
cp caminho/do/arquivo.txt caminho/novo/do/arquivo.txt

# Move-lo
mv caminho/do/arquivo.txt caminho/novo/do/arquivo.txt

# Renomea-lo (Atenção - para renomear um arquivo basta move-lo para a mesma pasta em que ele está inserido)
mv caminho/do/arquivo.txt caminho/do/arquivo_novo.txt

# Apagar o arquivo
rm arquivo_novo.txt

# Para ler um arquivo existe diversas opções
# O 'cat' vai exibir na tela principal todo o conteudo do arquivo
cat arquivo_novo.txt

# O 'less' vai abrir de forma mais discreta o arquivo
less arquivo_novo.txt

# Para exibir apenas o inicio do arquivo use
head arquivo_novo.txt

# Para exibir apenas o final do arquivo use
tail arquivo_novo.txt

# Para buscar por uma palavra dentro de um arquivo use a funcao 'grep'
grep "palavra buscada" arquivo.txt

# Buscar por um arquivo 
find /caminho/do/diretorio -name \nome_do_arquivo.txt
```

## Permissões de um arquivo
 
As permissões de um arquivo podem ser especificadas separadamente para “usuário”, “grupo” e “outros”. 
Os valores em octal são representados abaixo:

* 4 – leitura (r, de read).
* 2 – gravação (w, de write).
* 1 – execução (x, de execute)

Explicação: Para definir permissões, somam-se os valores acima. Por exemplo, para atribuir ao dono do arquivo (“usuário”) acesso total de leitura (r), gravação (w) e execução (x), basta somar o valor octal 4 + 2 + 1 = 7. Supondo que você queira limitar o acesso para membros do “grupo”, permitindo apenas a leitura e gravação, basta somar 4 + 2 = 6. 

Reunindo os dois exemplos citados, ficaria: chmod 760 (“r” para usuário, “w” para grupo e “0” para outros ou “rw-“).

Outros exemplos:
* chmod 777
o leitura (r), gravação (w) e execução (x) para todos (“usuário”, “grupo” e “outros”).
* chmod 755
o “rwx” para o “dono” (usuário), “rw” para o “grupo” e “outros”.

```bash
# Vamos usar os exemplos acima
# o leitura (r), gravação (w) e execução (x) para todos (“usuário”, “grupo” e “outros”)
chamod 777 arquivo.txt
ls -lh arquivo.txt

# o “rwx” para o “dono” (usuário), “rw” para o “grupo” e “outros”
chmod 755 arquivo.txt
ls -lh arquivo.txt

# utilizando a permissão +x para executar um arquivo .sh
# crie um arquivo .sh com o 'nano'
nano teste.sh

# dentro do arquivo escreva 
echo "ola mundo!"

# salve o arquivo, verifique as permissoes dele e em seguida dê permissão de execução
ls -lh texte.sh
chmod +x teste.sh
ls -lh texte.sh

# execute utilizando ./
./teste.sh

# exemplo if no arquivo teste.sh - existencia de arquivo!
```

## Verificando armazenamento dentro do terminal

Num sistema operacional Linux, monitorar o espaço em disco é crucial para manter o desempenho do sistema e evitar problemas de armazenamento insuficiente, como falhas em processos, perda de dados e serviços interrompidos.

O comando 'df': é usado para exibir a quantidade de espaço em disco disponível nos arquivos do sistema. Assim, é possível ter uma visão geral do espaço livre e ocupado nas partições montadas.
Algumas opções são muito úteis:
* -h. Exibe o espaço em formato legível para humanos, como KB, MB e GB.
* -a. Inclui dummy files.
* -T. Indica o tipo de sistema de arquivos.

O comando 'du': é usado para analisar o tamanho de diretórios e arquivos específicos.
Algumas opções são muito úteis:
* -h. Exibe o espaço em formato legível para humanos.
* -s. Resume o total para cada argumento.
* -a. Inclui arquivos e pastas.

O comando 'free': é usado para verificar a memoria ram.
Algumas opções são muito úteis:
* -h. Exibe o espaço em formato legível para humanos.

```bash
# verifique o espaço dos compartimentos usando df
df -h

# verifique o espaço dos diretorios usando du
du -sh /caminho/do/diretorio

# verifique o armazenamento de ram usando free
free -h
```

## Conda 
O conda é um gerenciador de pacotes, dependências e ambiente. Com ele é possível criar, manipular e alternar entre ambiantes. Em outras palavras, com o conda independentemente de onde você criar seu ambiente no final você vai conseguir replicar esse ambiente com as mesmas características, pois o conda é que ficará responsável por instalar os pacotes necessários para a sua aplicação. 

Para a instalação do conda, basicamente são necessários uma versão previamente instalada do python (2.7 ou 3.7). A forma mais fácil e rápida de se realizar a instalação é através da ferramenta Miniconda, que é composta basicamente do conda e suas dependências e pode ser baixada aqui: https://docs.conda.io/en/latest/ 



Após o download do software, são necessários rodar os seguintes comandos no terminal:

```bash
# Atribui-se a permissão de execução do script 
chmod +x Miniconda2-latest-Linux-x86_64.sh

# Execute o script 
./Miniconda2-latest-Linux-x86_64.sh

# Verifique a versão do seu conda 
conda --version
```

Em seguida você será capaz de utilizar as diversas funções do conda, como:
* criar um ambiente
* instalar pacotes nesse ambiente
* gerenciar os diversos ambientes criados

```bash
# Para criar um ambiente (ambiente de exemplo : aula_linux)
conda create --name aula_linux

# Para verificar todos os ambientes criados
conda info --envs

# Para ativar o ambiente criado 
conda activate aula_linux 
```

## Adicionando novos pacotes ao seu ambiente conda

É possível instalar pacotes de formas diversas dentro do ambiente conda. Os canais (ou Channels como são comumente chamados) são onde os pacotes de softwares ficam armazenados. Eles funcionam como repositórios de onde o Conda baixa e instala as bibliotecas e ferramentas. Ao executar o comando "install", o conda busca esse pacote nos canais cadastrados na sua configuração. 

Os canais mais comuns são: 
* Defaults: é o canal oficial mantido pelo Anaconda lnc. Ele contêm milhares de pacotes testeados e revisados divididos em categorias como 'main', 'r' e 'msys2'.
* Conda-forge: é o maior canal de mantido pela comunidade. Ele costuma ser mauito mais atualizado que o canal padrão e possui pacotes que não existem nos repositórios oficiais.
* Bioconda: é um canal voltado para a área de bioinformática e biologia computacional.

```bash
# Para instalar um pacote sem especificar um canal basta usar (usado como exemplo de pacote - numpy)
conda install numpy 

# Para instalar um pacote especificando o seu canal (usado como exemplo de pacote - bcftools)
conda install -c bioconda bcftools

# Você também pode adicionar um canal ao seu ambiente usando
conda config --add channels conda-forge

# Para visualizar quais canais estão presentes dentro do seu ambiente conda 
conda config --show channels


# Para visualizar quais pacotes estão instalados no seu ambiente 
conda list

# Para atualizar um pacote no conda
conda update bcftools

# Para remover um pacote do conda 
conda remove numpy
```

## Reprodutibilidade do trabalho

Ao compartilhar o seu trabalho com outras pessoas e ter garantia de que ele irá funcionar corretamente, é possível exportar o ambiente conda criado exclusivamente para esse trabalho. Isso garante que as versões das ferramentas utilizadas sejam as mesmas e evita erros de configuração. 

```bash
# Para exportar um ambiente conda 
conda env export > aula_linux.yml --name aula_linux

# Para a importação de um ambiente conda 
conda env create -f aula_linux.yml --name aula02
```

E por fim, encerrando um ambiente conda.

```bash
# Para desativar o ambiente. Primeiro volte para o "base"
conda deactivate

# Para remover, ou seja, excluir um ambiente 
conda remove --name aula_linux --all
```
